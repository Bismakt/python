```python
#first and last interchange.
s=input("Enter a string:")
print("New string is:",s[-1]+s[1:-1]+s[0])
```

```python
#create a string from first 2 and last 2 characters of a string.
s=input("Enter a string:")
print("New string is:",s[:2]+s[-2:])
```

```python
#create a single string separated with space from two strings by swapping the characters position 1.
s1=input("Enter first string:")
s2=input("Enter second string:")
print("New string is",s2[0]+s1[1:],s1[0]+s2[1:])
```

```python
#get a string from given string where all occurence of its first char have been changed to $,except the first char itself
s=input("Enter a string:")
print("New string is",s[0]+s[1:].replace(s[0],'$'))
```
