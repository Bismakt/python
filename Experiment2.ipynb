```python
#python prgram to convert celsius to fahrenheit(t(degree f)=t(degree c)*9/5+32)
celsius=float(input("Enter the temperature in celcius:"))
fahrnheit=celsius*9/5+32
print("Temperature in fahrenheit:",fahrnheit)
```

```python
#python program to convert kilometers to miles (1 kilometer is equal to 0.62137 miles)
k=float(input("Enter the distance in kilometer:"))
miles=k*0.62137
print(f"Distance in miles:{miles:.2f}")
```
