```python
%%writefile graphics/rectangle.py
#Rectangle module.
def area(l,b):
    return l*b
def perimeter(l,b):
    return 2*(l+b)
```

```python
%%writefile graphics/circle.py
#Circle module.
import math
def area(r):
    return math.pi*r*r
def perimeter(r):
    return 2*math.pi*r
```

```python
%%writefile graphics/graphics3D/cuboid.py
#Cube module.
def area(l,b,h):
    return 2*(l*b+l*h+b*h)
def perimeter(l,b,h):
    return 4*(l+b+h)
```

```python
%%writefile graphics/graphics3D/sphere.py
#Sphere module.
import math
def area(r):
    return 4*math.pi*r*r
def perimeter(r):
    return 2*math.pi*r
```

```python
#Create a package-area and perimeter of circle,rectangle,cuboid and sphere.
from graphics import rectangle,circle
from graphics.graphics3D import cuboid,sphere
while True:
    print("1.Rectangle\n2.Circle\n3.Cuboid\n4.Sphere\n5.Exit\n")
    ch=int(input("Enter your choice:"))
    if ch==1:
        l=int(input("Enter the length:"))
        b=int(input("Enter the breadth: "))
        print("Length:",l)
        print("Breadth:",b)
        print(f"Area of Rectangle:{rectangle.area(l,b):.2f}")
        print(f"Perimeter of Rectangle:{rectangle.perimeter(l,b):.2f}")
    elif ch==2:
        r=float(input("Enter the radius:"))
        print("Radius:",r)
        print(f"Area of Circle:{circle.area(r):.2f}")
        print(f"Perimeter of Circle:{circle.perimeter(r):.2f}")
    elif ch==3:
        l=int(input("Enter the length:"))
        b=int(input("Enter the breadth: "))
        h=int(input("Enter the height: "))
        print("Length:",l)
        print("Breadth:",b)
        print("Height:",h)
        print(f"Total surface area of Cuboid :{cuboid.area(l,b,h):.2f}")
        print(f"Perimeter of Cuboid:{cuboid.perimeter(l,b,h):.2f}")
    elif ch==4:
        r=float(input("Enter the radius:"))
        print("Radius:",r)
        print(f"Area of Sphere:{sphere.area(r):.2f}")
        print(f"Perimeter of Sphere:{sphere.perimeter(r):.2f}")
    elif ch==5:
        print("Exiting program.....")
        break
    else:
        print("Invalid choice")
```
