```python
#Display first and last colors.
color=input("Enter the colors:")
list=color.split(",")
print("First color:",list[0],"Second color:",list[-1])
```

```python
#print file extension
filename=input("Enter the filename:")
extension=filename.split(".")[-1]
print("Extension:",extension)
```

```python
#list the number and sum
number=input("Enter the numbers:")
lists=list(map(int,number.split(",")))
print("List is:",lists)
print("Sum is:",sum(lists))
```
